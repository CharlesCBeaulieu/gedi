```python
import numpy as np
good_cad_eig = [0.05036919325108474, 0.0009695007671234623, 0.00015837527042094853]
bad_cad_eig = [0.002393435861319129, 0.00007320627126247599, 0.000008511637156397139]

scan_eig = [0.043438662002055026, 0.0016824274375669905, 0.00018146679377800557]

def compute_score(eig1, eig2):
    eig1 = np.array(eig1)
    eig2 = np.array(eig2)
    
    print("scan", eig1)
    print("cad", eig2)
    

    diff = np.abs(eig1 - eig2)
    print("diff", diff)
    
    score = np.sum(diff)
    print("score", score)
    
    print("\n")
    return score

compute_score(scan_eig, good_cad_eig)
compute_score(scan_eig, bad_cad_eig)
```

```python
# load a pointcloud
import open3d as o3d
import numpy as np
import matplotlib.pyplot as plt
import os

pcd = o3d.io.read_point_cloud("bindmount/gedi_data_real_scan_5x315/scans/preprocess_point_cloud/000001.ply")
# get bbox 
bbox = pcd.get_axis_aligned_bounding_box()

# get the average distance between the nearest points
distances = []
for i in range(len(pcd.points)):
    distances.append(np.linalg.norm(np.array(pcd.points[i]) - np.array(bbox.get_center())))
distances = np.array(distances)
print("average distance to bbox center", np.mean(distances))
print(bbox)
```

```python
import os

def find_missing_files(folder1, folder2, extension=".ply"):
    """
    Compares two folders and returns:
      - files present in folder1 but missing in folder2,
      - files present in folder2 but missing in folder1.
    Only files ending with the given extension are considered.
    """
    files1 = {f for f in os.listdir(folder1) if f.endswith(extension)}
    files2 = {f for f in os.listdir(folder2) if f.endswith(extension)}
    
    missing_in_folder2 = files1 - files2
    missing_in_folder1 = files2 - files1
    
    return missing_in_folder2, missing_in_folder1

# Define the paths to your folders
folder_cads = "/app/bindmount/gedi_data_subset25_4m/cads/point_cloud"
folder_scans = "/app/bindmount/gedi_data_subset25_4m/scans/point_cloud"

print(len(os.listdir(folder_cads)), len(os.listdir(folder_scans)))

for f1, f2 in zip(os.listdir(folder_cads), os.listdir(folder_scans)):
    print(f1,f2)
missing_in_scans, missing_in_cads = find_missing_files(folder_cads, folder_scans)

# Print the results
if missing_in_scans:
    print("Files present in 'cads/point_cloud' but missing in 'scans/point_cloud':")
    for file in missing_in_scans:
        print(file)
else:
    print("No files are missing in 'scans/point_cloud' relative to 'cads/point_cloud'.")

if missing_in_cads:
    print("\nFiles present in 'scans/point_cloud' but missing in 'cads/point_cloud':")
    for file in missing_in_cads:
        print(file)
else:
    print("\nNo files are missing in 'cads/point_cloud' relative to 'scans/point_cloud'.")
```
